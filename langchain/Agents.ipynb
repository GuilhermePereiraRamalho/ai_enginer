```python
from langchain_openai import OpenAI
from langchain_experimental.agents.agent_toolkits import create_python_agent
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.prompts import PromptTemplate
import pandas as pd

from pathlib import Path
from decouple import AutoConfig
```

```python
cwd = Path.cwd()
project_dir = cwd / "langchain" if (cwd / "langchain").is_dir() else cwd
config = AutoConfig(search_path=str(project_dir))
api_key = config("OPENAI_API_KEY")
```

```python
openai = OpenAI(api_key=api_key, model_name="gpt-4o-mini", temperature=0)
```

```python
ddg_search = DuckDuckGoSearchRun()
```

```python
agent_executor = create_python_agent(
    llm=openai,
    tool=ddg_search,
    verbose=True
)
```

```python
prompt_template = PromptTemplate(
    input_variables=["query"],
    template="""
    Pesquise na web store {query} e forneça um resumo abrangente sobre o assunto.
    """
)
```

```python
query = "Carl Sagan"
prompt = prompt_template.format(query=query)
```

```python
response = agent_executor.invoke(prompt)
```

```python
print(f"Entrada do agente: {response['input']}")
```

```python
print(f"Saída do agente: {response['output']}")
```
