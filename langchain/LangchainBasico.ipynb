```python
from pathlib import Path
from langchain_openai import OpenAI, ChatOpenAI
from decouple import AutoConfig
```

```python
cwd = Path.cwd()
project_dir = cwd / "langchain" if (cwd / "langchain").is_dir() else cwd
config = AutoConfig(search_path=str(project_dir))
api_key = config("OPENAI_API_KEY")
```

```python
openai = OpenAI(api_key=api_key, model="gpt-4o-mini")
temperature = 1
frequency_penalty = 1
presence_penalty = 1
max_tokens = 500
n = 1
seed = 123

response = openai.invoke(
    input="Quem foi Carl Segan",
    temperature=temperature,
    frequency_penalty=frequency_penalty,
    presence_penalty=presence_penalty,
    max_tokens=max_tokens,
    n=n,
    seed=seed
)
print(response)
```

```python
openai = ChatOpenAI(api_key=api_key, model="gpt-4o-mini", store=False)

messages = [
    {"role": "system", "content": "Voce é um assistente que fornece informações sobre figuras históricas."},
    {"role": "user", "content": "Quem foi Carl Segan?"}
]

response = openai.invoke(messages)

print(response)
```
