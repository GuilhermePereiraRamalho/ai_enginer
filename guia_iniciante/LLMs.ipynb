```python
import transformers
from transformers import pipeline
```

1 - Gerador de texto

```python
gerador = pipeline("text-generation", model="pierreguillou/gpt2-small-portuguese")
```

```python
texto = "Em sentido estrito, ciência refere-se ao sistema de adquirir conhecimento baseado no método científico."
resultado = gerador(
    texto,
    max_length=60,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    top_p=0.85
)
print(resultado)
```

```python
print(resultado[0]['generated_text'])
```

```python
print(resultado)
```

2 - Preencher Lacunas

```python
mascarar = pipeline("fill-mask", model="neuralmind/bert-base-portuguese-cased")
```

```python
texto = mascarar("Existe uma chance do copo cair no [MASK]")
for x in range(len(texto)):
  print(texto[x])
```

```python
texto = mascarar(["Existe uma chance do copo cair no [MASK]", "Brasilia é a capital do [MASK]"])
for x in range(len(texto)):
  print(texto[x])
```

3 - Perguntas e respostas

```python
qea = pipeline("question-answering", model="pierreguillou/bert-base-cased-squad-v1.1-portuguese")
```

```python
texto = "Carl Segan foi um cientista norte-americano. Segan é autor de mais de 600 publicações científicas e tam'bem de mais de vinte livros de ciência e ficção científica"
pergunta = "Quantas publicações científicas Carl Segan tem publicado?"
resposta = qea(question=pergunta, context=texto)
print(f"Pergunta: {pergunta}")
print(f"Resposta: {resposta['answer']}")
print(f"Score: {resposta['score']}")
```

4 - Resumo

```python
resumidor = pipeline("summarization")
```

```python
texto = """
Carl Edward Sagan foi um cientista planetário, astrônomo, astrobiólogo, astrofísico,
escritor, divulgador científico e ativista norte-americano. Sagan é autor de
mais de 600 publicações científicas e também de mais de vinte livros de ciência
e ficção científica. Foi durante a vida um grande defensor do ceticismo e do uso
do método científico. Promoveu a busca por inteligência extraterrestre através
do projeto SETI e instituiu o envio de mensagens a bordo de sondas espaciais,
destinadas a informar possíveis civilizações extraterrestres sobre a existência
humana. Mediante suas observações da atmosfera de Vênus, foi um dos primeiros
cientistas a estudar o efeito estufa em escala planetária. Também fundou a
organização não governamental Sociedade Planetária e foi pioneiro no ramo da
exobiologia. Sagan passou grande parte da carreira como professor da Universidade
Cornell, onde foi diretor do laboratório de estudos planetários. Em 1960 obteve
o título de doutor pela Universidade de Chicago.
"""
```

```python
resumo = resumidor(texto, max_length=100, min_length=50)
print(resumo)
```
