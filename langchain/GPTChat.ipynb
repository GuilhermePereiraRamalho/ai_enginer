```python
from pathlib import Path
from openai import OpenAI
from decouple import AutoConfig
```

```python
cwd = Path.cwd()
project_dir = cwd / "langchain" if (cwd / "langchain").is_dir() else cwd
config = AutoConfig(search_path=str(project_dir))
api_key = config("OPENAI_API_KEY")
```

```python
client = OpenAI(api_key=api_key)
model = "gpt-4o-mini"
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "user", "content": "conte uma piada!"}
    ],
    store=False
)
```

```python
print(response)
```

```python
print(type(response))
```

Chat Basico

```python
message = (response.choices[0].message.content)
print(message)
```

Roles

```python
response = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "system", "content": "You are a fictional investment assistant."},
        {"role": "user", "content": "What is the best low-risk investment you recommend for this year?"},
    ],
    store=False
)
```

```python
message = (response.choices[0].message.content)
print(message)
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "assistant", "content": message},
        {"role": "user", "content": "What are the risks related to treasury bonds?"},
    ],
    store=False
)
```

```python
print(response.choices[0].message.content)
```

Hiper Parametros

```python
response = client.chat.completions.create(
    model=model,
    frequency_penalty=1,
    presence_penalty=1,
    temperature=1,
    max_tokens=500,
    n=2,
    seed=123,
    store=False,
    # stop=["shadows", "mortal "],
    messages=[
        {"role": "system", "content": "You are a depressed and disillusioned poet."},
        {"role": "user", "content": "Compose a poem about the ephemerality of existence."},
    ],
)
```

```python
print(response.choices[0].message.content)
print("----------------------------------")
print(response.choices[1].message.content)
```
